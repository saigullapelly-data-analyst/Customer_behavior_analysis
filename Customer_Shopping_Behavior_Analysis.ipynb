```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.columns
```

```python
df.columns = df.columns.str.replace('purchase_amount_(usd)_(usd)','purchase_amount')
df.columns
```

```python
df = df.rename(columns = {'purchase_amount_(usd)': 'purchase_amount'})
df.columns
```

```python
df.columns
```

```python
labels = ['Young Adult','Adult','Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'] ,q=4,labels =labels)
df[['age','age_group']]
df.columns
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency_mapping = {
    'Fortnightly' : 14, 'Weekly' : 7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly':14,
       'Monthly':30, 'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']]
```
